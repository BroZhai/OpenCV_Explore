```python

# define a function and recall it to process the string (delete the same ): Remove the same characters and sort the remaining characters.

#input: s = “ajldjlajfdljfddd”--> output: ”adfjl”



def test(s):
    str_list = []

    for i in s:
        if i in str_list:
            str_list.remove(i)

        str_list.append(i)
    a = sorted(str_list)
    return "".join(a)


if __name__=='__main__':
    s='ajldjlajfdljfddd'
    output=test(s)
    print(output)
    
```
